```python
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
import json

mpl.rcParams['xtick.labelsize'] = 20
mpl.rcParams['ytick.labelsize'] = 20
mpl.rcParams['font.size'] = 20
mpl.rcParams['figure.autolayout'] = False
mpl.rcParams['axes.linewidth'] = 3
mpl.rcParams['axes.titlesize'] = 20
mpl.rcParams['axes.labelsize'] = 20
mpl.rcParams['lines.linewidth'] = 2.2
mpl.rcParams['lines.markersize'] = 5
mpl.rcParams['lines.markeredgewidth'] = 1.2
mpl.rcParams['errorbar.capsize'] = 5
mpl.rcParams['xtick.major.width'] = 3
mpl.rcParams['xtick.major.size'] = 10
mpl.rcParams['ytick.major.width'] = 3
mpl.rcParams['ytick.major.size'] = 8
mpl.rcParams['xtick.minor.width'] = 1.5
mpl.rcParams['xtick.minor.size'] = 6
mpl.rcParams['ytick.minor.width'] = 1.5
mpl.rcParams['ytick.minor.size'] = 6
mpl.rcParams['legend.fontsize'] = 14
mpl.rcParams['mathtext.fontset'] = 'dejavusans'
mpl.rcParams['font.family'] = 'sans-serif'
cm_colors = plt.get_cmap('Set2')
```

```python
method = "TQA_PP"
colors = plt.get_cmap("Set1")
with open(f"{method}.json", "r") as file:
    info = json.load(file)
fig, ax = plt.subplots()
ax.plot(info["exact"].keys(), info["exact"].values(), marker="o", markersize=8, markeredgecolor="black", linestyle=":", color=colors(0))
ax.plot(info["approx"].keys(), info["approx"].values(), marker="o", label="PP", markersize=8, markeredgecolor="black", color=colors(0))
method = "TQA_MPS"
with open(f"{method}.json", "r") as file:
    info = json.load(file)
ax.plot(info["exact"].keys(), info["exact"].values(), marker="o", markersize=8, markeredgecolor="black", linestyle=":", color=colors(1))
ax.plot(info["approx"].keys(), info["approx"].values(), marker="o", label="MPS", markersize=8, markeredgecolor="black", color=colors(1))
ax.plot([],[], color="black", linestyle=":",label="SV")
ax.set_xlabel("p")
ax.set_ylabel("energy")
ax.set_title("TQA")
ax.legend()
fig.savefig(f"comparison_TQA.pdf", bbox_inches="tight", transparent=True)
```
