```python
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
import json

cm_colors = plt.get_cmap('Set2')
```

```python
fig, axs = plt.subplots(2, 1, figsize=(4.5, 5))

method = "FA_PP"
colors = plt.get_cmap("Set1")

with open(f"{method}.json", "r") as file:
    info = json.load(file)

axs[0].plot(info["exact"].keys(), info["exact"].values(), marker="s", markersize=6, markeredgecolor="black", linestyle=":", color="#4477AA")
axs[0].plot(info["approx"].keys(), info["approx"].values(), marker="s", label="PP", markersize=6, markeredgecolor="black", color="#4477AA")
method = "FA_MPSAer"

with open(f"{method}.json", "r") as file:
    info = json.load(file)

axs[0].plot(info["exact"].keys(), info["exact"].values(), marker="^", markersize=6, markeredgecolor="black", linestyle=":", color="#4477AA")
axs[0].plot(info["approx"].keys(), info["approx"].values(), marker="^", label="MPS (Aer)", markersize=6, markeredgecolor="black", color="#4477AA")
axs[0].plot([],[], color="black", linestyle=":",label="SV")
axs[0].set_xlabel("p")
axs[0].set_ylabel("energy")
#axs[0].set_title("FA")
axs[0].legend(loc=4)

method = "TQA_PP"

with open(f"{method}.json", "r") as file:
    info = json.load(file)

axs[1].plot(info["exact"].keys(), info["exact"].values(), marker="s", markersize=6, markeredgecolor="black", linestyle=":", color="#AA3377")
axs[1].plot(info["approx"].keys(), info["approx"].values(), marker="s", label="PP", markersize=6, markeredgecolor="black", color="#AA3377")
method = "TQA_MPSAer"
with open(f"{method}.json", "r") as file:
    info = json.load(file)
axs[1].plot(info["exact"].keys(), info["exact"].values(), marker="^", markersize=6, markeredgecolor="black", linestyle=":", color="#AA3377")
axs[1].plot(info["approx"].keys(), info["approx"].values(), marker="^", label="MPS", markersize=6, markeredgecolor="black", color="#AA3377")
axs[1].plot([],[], color="black", linestyle=":",label="SV")

for ax in axs:
    ax.set_xlabel("QAOA depth")
    ax.set_ylabel("Energy")
#ax.set_title("TQA")
#axs[1].legend()

axs[0].text(0.01, 23, "(a) Fixed angles$^\star$")
axs[1].text(0.01, 21.45, "(b) TQA$^\star$")

fig.tight_layout()

fig.savefig(f"comparison_MPS_PP_SV.pdf", bbox_inches="tight", transparent=True)
```
