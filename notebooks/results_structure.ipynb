# Parameter setting methods results

In the following NB we show what the structure of the results from the QAOA parameter training looks like. The training is run with the following command

```python
python -m train 
--input <PATH>/instances/erdos_renyi/000_10nodes_erdosrenyi20percent.json 
--config <PATH>/methods/example_method.json 
--save 
--save_dir <PATH>/data/examples/ 
--save_file example.json 
--pre_factor -0.5
```

This loads the training method from `example_method.json` and saves the result of the parameter training under the folder `data/examples/`. Now, we will take a look at the structure of this data.

````python
import json
import pprint as pp
````

## Training method

First, we show the method used to train the parameters. Here, the training was done for a depth-one QAOA with SciPy initialized from a random point.

````python
with open("../methods/example_method.json", "r") as fin:
    method = json.load(fin)

pp.pprint(method)
````

## Result structure

The `train.py` saves the result of the parameter training in a file. We now show the contents of this file as it includes more than simply the optimized QAOA parameters.

````python
with open("../data/examples/20250721_171511_example.json", "r") as fin:
    data = json.load(fin)
````

First, the data has two keys to represent the trainers that were used.

````python
print(data.keys())
````

The first training, i.e., `0` is just a random point and therefore does not contain much data. The data shows some system information too.

````python
data["0"]
````

The second trainer is the actual SciPy optimization which includes the training history. Not also that the options given to the SciPy trainer are stored in the output.

````python
data["1"]
````
