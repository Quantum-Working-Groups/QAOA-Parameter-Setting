# Missing Instances


```python
def number_to_run(
    evaluation: str,
    graph_type: str,
    method: str,
    methods_to_exclude: list[str] | None = None,
):

    num_instances_for_config = len(
        table.get_graph_instances_for_config(
            trainer_method=method,
            graph_type=graph_type,
            depth=10,
            evaluation_method=evaluation,
        )
    )
    if num_instances_for_config == 0:
        # If we have no instances for this specific configuration, it is a trivial reference and we
        # never want to select it.
        return 9999999999

    df = table.get_missing_instances_summary(
        reference_methods={(evaluation, graph_type): method},
        # {
        #     ("SV", "erdos_renyi"): "LR_opt.json",
        #     ("SV", "heavy_hex"): "LR_opt.json",
        #     ("SV", "random_regular"): "TQA_opt.json",
        #     ("SV", "line_to_full"): "LR_opt.json",
        # },
        depth=10,
        methods_to_exclude=methods_to_exclude,
    )

    def _map_method(m: str) -> str:
        _m = m.lower()
        if "no_opt" in _m:
            return m.replace("no_opt", "noOpt")
        return m

    if len(df) == 0:
        # Early return because there are no missing instances.
        return 0

    df["trainer_method"] = df["trainer_method"].apply(_map_method)

    # Filter methods where there is an opt and a no-opt where the no-opt is derived from the zeroth
    # iteration of the opt.
    all_opt_methods = [
        method
        for method in df["trainer_method"].unique()
        if not method.lower().endswith("_noopt.json")
    ]
    all_methods = df["trainer_method"].unique()
    keep_methods = set(all_opt_methods)
    for method in all_methods:
        if method in all_opt_methods:
            pass
        if not ("TQA" in method or "FA" in method or "LR" in method):
            continue
        no_opt_method = method
        opt_method = method.replace("noOpt", "opt")
        if opt_method not in all_methods:
            # We cannot find an opt method that matches this no-opt, so we keep the no-opt.
            # print(
            #     f"Cannot find opt method {opt_method!r} for no-opt method {no_opt_method!r}"
            # )
            keep_methods.add(no_opt_method)
        else:
            # We can find the opt method that matches this no-opt, so we only keep the opt.
            # print(
            #     f"Found opt-method {opt_method!r} for no-opt method {no_opt_method!r}"
            # )
            keep_methods.add(opt_method)
    # print(f"Methods removed: {set(all_methods) - set(keep_methods)!r}")
    return int(df[df["trainer_method"].isin(keep_methods)]["num_missing"].sum())
```

```python
reference_methods: dict[tuple[str, str], str] = {}
prev_evaluation = "Nothing"
for evaluation, graph_type, methods_to_exclude in [
    # 1. MPS
    (
        "MPS",
        "erdos_renyi",
        [],
    ),
    (
        "MPS",
        "heavy_hex",
        #
        ["FAAer_no_opt.json", "FAAer_opt.json"],
    ),
    (
        "MPS",
        "line_to_full",
        [
            "FAAer_no_opt.json",
            "FAAer_opt.json",
            "FAer.json",
            "IAer.json",
            "TQAAer_no_opt.json",
            "TQAAer_opt.json",
        ],
    ),
    (
        "MPS",
        "random_regular",
        [],
    ),
    # 2. PP
    (
        "PP",
        "erdos_renyi",
        [
            "FA_opt.json",
            "FA_no_opt.json",
            "FA_no_opt.json",
            "LR_no_opt.json",
            "LR_opt.json",
        ],
    ),
    (
        "PP",
        "heavy_hex",
        ["FA_opt.json", "FA_no_opt.json", "LR_no_opt.json", "LR_opt.json"],
    ),
    (
        "PP",
        "line_to_full",
        ["FA_opt.json", "FA_no_opt.json"],
    ),
    (
        "PP",
        "random_regular",
        ["LR_opt.json", "LR_no_opt.json"],
    ),
    # 3. SV evaluation method
    ("SV", "erdos_renyi", ["FA_opt.json", "FA_noOpt.json", "F.json"]),
    (
        "SV",
        "heavy_hex",
        ["FA_opt.json", "FA_noOpt.json", "TQA_noOpt.json", "TQA_opt.json"],
    ),
    ("SV", "line_to_full", ["FA_opt.json", "FA_noOpt.json"]),
    ("SV", "random_regular", ["FA_opt.json", "FA_noOpt.json"]),
]:
    if evaluation != prev_evaluation:
        print("=" * 20)
        prev_evaluation = evaluation
    best_method: str | None = None
    best_method_count = 99999999

    def exclude_if_mps(method: str) -> bool:
        if evaluation != "MPS":
            return False
        # print(method)
        if "Aer" not in method:
            return True
        return False

    methods_to_sweep = [
        table.trainer_config_to_method(method)
        for method in df[(df["evaluation"] == evaluation) & (df["depth"] == 10)][
            "trainer_config"
        ].unique()
        if table.trainer_config_to_method(method) not in methods_to_exclude
        and not exclude_if_mps(table.trainer_config_to_method(method))
    ]
    # Always exclude transition states, RTS and TS.
    methods_to_exclude = ["RTS.json", "TS.json", "PT_AAAM.json"]
    if evaluation == "MPS":
        # If we are dealing with MPS, remove non-Aer results.
        methods_to_exclude.extend(
            [
                table.trainer_config_to_method(method)
                for method in df[
                    (df["evaluation"] == evaluation) & (df["depth"] == 10)
                ]["trainer_config"].unique()
                if table.trainer_config_to_method(method) not in methods_to_exclude
                and exclude_if_mps(table.trainer_config_to_method(method))
            ]
        )
    for method in methods_to_sweep:
        # Always ignore RTS and TS
        if "RTS" in method or "TS" in method:
            continue
        num = number_to_run(
            evaluation, graph_type, method, methods_to_exclude=methods_to_exclude
        )
        if num < best_method_count:
            best_method_count = num
            best_method = method
    if best_method is None:
        raise RuntimeError(
            f"Could not find best method for {graph_type!r} and {evaluation!r}."
        )
    print(
        f"Best eval={evaluation!r} and graph_type={graph_type!r} is {best_method!r} with {best_method_count} new experiments."
    )
    reference_methods[(evaluation, graph_type)] = best_method
pivot, styler, _ = formatted_styler_for(
    table,
    depths=10,
    num_nodes=None,
    agg_values="num_instances",
    with_fancy_values=True,
    # cmap="Greens",
    cmap=[{k: v} for k, v in {"MPS": "YlOrBr", "PP": "YlGn", "SV": "PuBu"}.items()],
    precision=0,
    missing_data_str="-",
    target_format="text",
    show_empty_rows=False,
    reference_methods=reference_methods,
    exclude_methods=["TS", "RTS", "PT_AAAM"],
    restrict_mps_to_aer=True,
)
styler
```

```python
def has_equivalent_opt(method: str) -> bool:
    no_opt_prefixes = ["TQA", "FA", "FAAer", "TQAAer", "LR", "LRAer"]
    return any(prefix not in method.split("_") for prefix in no_opt_prefixes) and (
        "noOpt" in method or "no_opt" in method
    )


def get_equivalent_opt(method: str) -> str:
    return method.replace("noOpt", "opt").replace("no_opt", "opt")
```

```python
missing_instances: dict[tuple[str, str, str], set[str]] = (
    table.get_missing_instances_after_filter(
        reference_methods=reference_methods, depth=10
    )
)
missing_records: list[dict[str, str]] = []
num_extra_simulations = 0
for (__evaluation, __graph_type, __method), __instances in missing_instances.items():
    # Check if the method is an Aer method. If not, skip this entry.
    if __evaluation == "MPS":
        if "Aer" not in __method:
            # print(
            #     f"Ignoring {__evaluation}, {__graph_type}, {__method}: Not Aer method for MPS evaluation"
            # )
            continue
    # Ignore RTS, TS, and PT_AAAM
    if any(
        ignore in __method.split(".")[0].split("_") for ignore in ["RTS", "TS", "AAAM"]
    ):
        # print(f"Ignoring {__evaluation}, {__graph_type}, {__method}: Ignore method")
        continue

    num_extra_simulations += len(__instances)
    if has_equivalent_opt(__method):
        __opt_method = get_equivalent_opt(__method)
        if (__evaluation, __graph_type, __opt_method) in missing_instances:
            # print(
            #     f"Ignoring {__evaluation}, {__graph_type}, {__method}: Zeroth iteration of {__opt_method!r}"
            # )
            continue

    print(f"{__evaluation}\t{__method: <20}\t{__graph_type}\t{len(__instances)}")
    for _inst in sorted(__instances):
        print(f"  - {_inst}")
        missing_records.append(
            {"evaluation": __evaluation, "method": __method, "instance": _inst}
        )
print(f"Number of new experiments: {num_extra_simulations}")
```
